Q1. String operations
Write a function that returns the reversed string and the characters at every second position (2nd, 4th, 6th, ...).

```python


def string_operations(text):
    reversed_text = text[::-1]
    every_second_character = text[1::2]
    return reversed_text, every_second_character

result = string_operations("Computer")
print("Reversed string:", result[0])
print("Every 2nd character:", result[1])
```

Q2. Squares of even numbers
Use list comprehension to create a new list containing only the squares of even numbers.

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even_squares = [number ** 2 for number in numbers if number % 2 == 0]

print("Original list:", numbers)
print("Squares of even numbers:", even_squares)
```

Q3. Character frequency
Write a function that counts how many times each character appears in a string using a dictionary.

```python
def character_frequency(text):
    frequency = {}
    for character in text:
        frequency[character] = frequency.get(character, 0) + 1
    return frequency

text = "programming"
print("Character frequency:", character_frequency(text))
```

Q4. Create a dictionary using zip()
Write a function that takes two lists and combines them into a dictionary using zip().

```python
def lists_to_dictionary(keys, values):
    return dict(zip(keys, values))

subjects = ["AI", "Python", "C++"]
marks = [90, 85, 88]

result = lists_to_dictionary(subjects, marks)
print("Dictionary:", result)
```

Q5. Manual iteration using iter() and next()
Use iter() and next() to print list elements one by one. Handle StopIteration with try-except.

```python
items = ["Python", "C++", "Git", "GitHub"]
iterator = iter(items)

while True:
    try:
        item = next(iterator)
        print(item)
    except StopIteration:
        print("StopIteration handled: no more items.")
        break
```

Q6. Palindrome check using slicing
Write a function that checks whether a string is a palindrome using slicing.

```python
def is_palindrome(text):
    return text == text[::-1]

word = "madam"
if is_palindrome(word):
    print(word, "is a palindrome.")
else:
    print(word, "is not a palindrome.")
```

Q7. Second largest number without sorting
Write a function that finds the second largest distinct number in a list without using sorting.

```python
def second_largest(numbers):
    largest = None
    second = None

    for number in numbers:
        if largest is None or number > largest:
            second = largest
            largest = number
        elif number != largest and (second is None or number > second):
            second = number

    if second is None:
        return "A second distinct largest number does not exist."
    return second

values = [12, 35, 1, 10, 34, 35]
print("List:", values)
print("Second largest number:", second_largest(values))
```
